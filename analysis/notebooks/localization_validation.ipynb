# Localization validation

Loads a pose log written by `PoseLogger` (see `TeamCode/POSE_LOGGING.md` for how to pull
the CSV off the Control Hub) and compares the logged path against a simulated/planned one.

CSV columns: `time_s, x_in, y_in, heading_deg`.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

LOG_DIR = Path("../../logs")

log_files = list(LOG_DIR.glob("*.csv"))

if not log_files:
    raise FileNotFoundError(f"No CSV log files found in {LOG_DIR.resolve()}")

LOG_PATH = max(log_files, key=lambda p: p.stat().st_mtime)

print(f"Loading: {LOG_PATH}")

log = pd.read_csv(LOG_PATH)
log.head()
```

```python
fig, ax = plt.subplots(figsize=(6, 6))
ax.plot(log["x_in"], log["y_in"], marker=".", label="logged")
ax.set_xlabel("x (in)")
ax.set_ylabel("y (in)")
ax.set_title("Logged robot path")
ax.axis("equal")
ax.legend()
plt.show()
```

```python
fig, axes = plt.subplots(3, 1, figsize=(8, 6), sharex=True)
axes[0].plot(log["time_s"], log["x_in"])
axes[0].set_ylabel("x (in)")
axes[1].plot(log["time_s"], log["y_in"])
axes[1].set_ylabel("y (in)")
axes[2].plot(log["time_s"], log["heading_deg"])
axes[2].set_ylabel("heading (deg)")
axes[2].set_xlabel("time (s)")
plt.tight_layout()
plt.show()
```
